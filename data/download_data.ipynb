```python
import os
import urllib.request
import tarfile

# NOTE
# Spawrious provides separate folders for O2O and M2M splits,
# but the underlying images used for O2O are a subset of the M2M data.
# Therefore downloading both leads to duplicated images.
# To avoid having duplicate data, this implementation uses only the M2M folders
# as the physical data source. All splits are defined logically via metadata.

download_spawrious_o2o_data = False
```

```python
waterbirds_url = "https://nlp.stanford.edu/data/dro/waterbird_complete95_forest2water2.tar.gz"
data_dir = "waterbird_complete95_forest2water2"

if not os.path.exists("waterbirds.tar.gz"):
    urllib.request.urlretrieve(waterbirds_url, "waterbirds.tar.gz")
with tarfile.open("waterbirds.tar.gz", "r:gz") as tar:
    tar.extractall()
os.remove("waterbirds.tar.gz")
```

```python
waterbirds_url = "https://data.caltech.edu/records/w9d68-gec53/files/segmentations.tgz?download=1"
if not os.path.exists(os.path.join("waterbird_complete95_forest2water2", "segmentations", "segmentations.tgz")):
    urllib.request.urlretrieve(waterbirds_url, os.path.join("waterbird_complete95_forest2water2", "segmentations", "segmentations.tgz"))

with tarfile.open(os.path.join("waterbird_complete95_forest2water2", "segmentations", "segmentations.tgz"), "r:gz") as tar:
    tar.extractall()
```
